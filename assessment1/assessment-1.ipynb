```python
# Task 6: Adding HDI Category Column
def classify_hdi(hdi_value):
    if hdi_value < 0.550:
        return 'Low'
    elif 0.550 <= hdi_value <= 0.699:
        return 'Medium'
    elif 0.700 <= hdi_value <= 0.799:
        return 'High'
    else:  # >= 0.800
        return 'Very High'

hdi_2022_df['HDI_Category'] = hdi_2022_df['hdi'].apply(classify_hdi)
```

Comparative Regional Analysis: South Asia vs Middle East

```python
df_recent = df[df['year'].isin([2020, 2021, 2022])].copy()
```

```python
# Task 1: Define regions and create subsets
south_asia_countries = ["Afghanistan", "Bangladesh", "Bhutan", "India", "Maldives", "Nepal", "Pakistan", "Sri Lanka"]

middle_east_countries = ["Bahrain", "Iran", "Iraq", "Israel", "Jordan", "Kuwait", "Lebanon", "Oman", 
                         "Palestine", "Qatar", "Saudi Arabia", "Syria", "United Arab Emirates", "Yemen"]

south_asia = df_recent[df_recent['country'].isin(south_asia_countries)].copy()
middle_east = df_recent[df_recent['country'].isin(middle_east_countries)].copy()
```

```python
print("South Asia countries in data:", sorted(south_asia['country'].unique()))
print("Middle East countries in data:", sorted(middle_east['country'].unique()))
print("South Asia observations:", south_asia.shape[0])
print("Middle East observations:", middle_east.shape[0])
```

```python
# Save subsets
south_asia.to_csv('HDI_SouthAsia_2020_2022.csv', index=False)
middle_east.to_csv('HDI_MiddleEast_2020_2022.csv', index=False)
print("\nSubsets saved as 'HDI_SouthAsia_2020_2022.csv' and 'HDI_MiddleEast_2020_2022.csv'")
```

```python
# Task 2: Descriptive Statistics for HDI
print("\n--- Descriptive Statistics (HDI 2020-2022) ---")
sa_mean = south_asia['hdi'].mean()
sa_std = south_asia['hdi'].std()
me_mean = middle_east['hdi'].mean()
me_std = middle_east['hdi'].std()

print(f"South Asia - Mean HDI: {sa_mean:.4f}, Std: {sa_std:.4f}")
print(f"Middle East - Mean HDI: {me_mean:.4f}, Std: {me_std:.4f}")
print("Better performing region on average: " + ("Middle East" if me_mean > sa_mean else "South Asia"))
```

```python
# Task 3: Top and Bottom Performers
# Use average HDI over 2020-2022 per country
sa_avg = south_asia.groupby('country')['hdi'].mean().sort_values(ascending=False)
me_avg = middle_east.groupby('country')['hdi'].mean().sort_values(ascending=False)

print("\nSouth Asia Top 3:", sa_avg.head(3).index.tolist())
print("South Asia Bottom 3:", sa_avg.tail(3).index.tolist())
print("Middle East Top 3:", me_avg.head(3).index.tolist())
print("Middle East Bottom 3:", me_avg.tail(3).index.tolist())
```

```python
# Bar chart for top/bottom
top_bottom = pd.concat([
    sa_avg.head(3), sa_avg.tail(3),
    me_avg.head(3), me_avg.tail(3)
])

plt.figure(figsize=(10, 6))
top_bottom.plot(kind='bar')
plt.title('Top 3 and Bottom 3 Countries by Average HDI (2020-2022)')
plt.ylabel('Average HDI')
plt.xlabel('Country')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Task 4: Metric Comparisons (Grouped Bar Charts)
metrics = {
    'Gender Development Index': 'gender_development',
    'Life Expectancy': 'life_expectancy',
    'GNI per Capita': 'gross_inc_percap'
}

for metric_name, col in metrics.items():
    sa_val = south_asia[col].mean()
    me_val = middle_east[col].mean()
    disparity = abs(sa_val - me_val)
    
    print(f"\n{metric_name} - South Asia: {sa_val:.4f}, Middle East: {me_val:.4f}, Disparity: {disparity:.4f}")
    
    # Grouped bar
    pd.DataFrame({
        'South Asia': [sa_val],
        'Middle East': [me_val]
    }).plot(kind='bar', title=f'Average {metric_name} (2020-2022)')
    plt.ylabel('Value')
    plt.xticks(rotation=0)
    plt.legend()
    plt.tight_layout()
    plt.show()

   # Greatest disparity (manually from prints or compute)
print("\nIdentify the metric with highest disparity from the values above.")
```

```python
# Task 5: HDI Disparity
sa_range = south_asia['hdi'].max() - south_asia['hdi'].min()
me_range = middle_east['hdi'].max() - middle_east['hdi'].min()
sa_cv = sa_std / sa_mean
me_cv = me_std / me_mean

print(f"\nHDI Range - South Asia: {sa_range:.4f}, Middle East: {me_range:.4f}")
print(f"HDI CV - South Asia: {sa_cv:.4f}, Middle East: {me_cv:.4f}")
print("Region with more variation: " + ("Middle East" if me_cv > sa_cv else "South Asia"))
```

```python
# Task 6: Correlation Analysis
regions = {'South Asia': south_asia, 'Middle East': middle_east}

for name, data in regions.items():
    print(f"\n--- {name} Correlations ---")
    corr_gdi = data['hdi'].corr(data['gender_development'])
    corr_le = data['hdi'].corr(data['life_expectancy'])
    print(f"Correlation HDI vs Gender Development: {corr_gdi:.4f}")
    print(f"Correlation HDI vs Life Expectancy: {corr_le:.4f}")
    
    # Scatter plots with trendline
    fig, ax = plt.subplots(1, 2, figsize=(12, 5))
    
    sns.scatterplot(x='gender_development', y='hdi', data=data, ax=ax[0])
    sns.regplot(x='gender_development', y='hdi', data=data, scatter=False, ax=ax[0], color='red')
    ax[0].set_title(f'{name}: HDI vs Gender Development')
    
    sns.scatterplot(x='life_expectancy', y='hdi', data=data, ax=ax[1])
    sns.regplot(x='life_expectancy', y='hdi', data=data, scatter=False, ax=ax[1], color='red')
    ax[1].set_title(f'{name}: HDI vs Life Expectancy')
    
    plt.tight_layout()
    plt.show()

print("\nInterpretation: Strong positive correlations expected, especially with life expectancy. Closer to 1 indicates stronger relationship.")
```

```python
# Task 7: Outlier Detection (1.5 IQR rule)
for name, data in regions.items():
    print(f"\n--- Outliers in {name} ---")
    
    for var in ['hdi', 'gross_inc_percap']:
        Q1 = data[var].quantile(0.25)
        Q3 = data[var].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        outliers = data[(data[var] < lower) | (data[var] > upper)]
        
        print(f"{var} outliers ({len(outliers)}): {outliers['country'].tolist() if not outliers.empty else 'None'}")
        
        # Scatter plot highlighting outliers
        plt.figure(figsize=(8, 6))
        sns.scatterplot(x='gross_inc_percap', y='hdi', data=data, label='Normal')
        if not outliers.empty:
            sns.scatterplot(x='gross_inc_percap', y='hdi', data=outliers, color='red', label='Outlier')
        plt.title(f'{name}: HDI vs GNI per Capita (Outliers in red)')
        plt.xlabel('GNI per Capita')
        plt.ylabel('HDI')
        plt.legend()
        plt.tight_layout()
        plt.show()

print("\nDiscussion: Outliers often high-income Gulf states (high GNI, variable HDI) or conflict-affected countries (low HDI). They highlight inequalities within regions.")
```
